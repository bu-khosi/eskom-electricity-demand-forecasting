```python
#DEVELOPING FORECASTING MODELS

#LINEAR REGRESSION
#RANDOM FOREST
#XGBOOST

#We'll use an 80% chronological training set and a 20% testing set. This ensures the models are trained on earlier observations and evaluated on later observations

# Define the target variable
y = feature_df['Residual Demand']

# Define the predictor variables
X = feature_df.drop(columns=['Residual Demand'])

print("Features shape:", X.shape)
print("Target shape:", y.shape)

```

```python
# MODEL EVAULATION


We will compare the three models using:
MAE:MEAN ABSOLUTE ERROR (Average absolute prediction error.)
RMSE:ROOT MEAN SQUARE ERROR (Penalises larger prediction errors more heavily.)
MAPE: Average percentage prediction error.
Lower values indicate better performance for all three metrics
```

```python
#CALCULATE EVALUATION  METRICS

from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    mean_absolute_percentage_error
)

import numpy as np
import pandas as pd

# Check that Linear Regression was trained
if 'lr_predictions' not in globals():
    lr_model.fit(X_train, y_train)
    lr_predictions = lr_model.predict(X_test)

# Calculate metrics for each model
models = {
    'Linear Regression': lr_predictions,
    'Random Forest': rf_predictions,
    'XGBoost': xgb_predictions
}

results = []

for model_name, predictions in models.items():

    mae = mean_absolute_error(y_test, predictions)

    rmse = np.sqrt(
        mean_squared_error(y_test, predictions)
    )

    mape = mean_absolute_percentage_error(
        y_test, predictions
    ) * 100

    results.append({
        'Model': model_name,
        'MAE': mae,
        'RMSE': rmse,
        'MAPE (%)': mape
    })

results_df = pd.DataFrame(results)

print("MODEL EVALUATION RESULTS")
display(results_df.round(3))
```

```python
#IDENTIFYING THE BEST WORKING/PERFOMING MODEL


# Rank models by their evaluation metrics
ranked_results = results_df.sort_values(
    by='RMSE',
    ascending=True
)

print("Models ranked by RMSE:")
display(ranked_results.round(3))

best_model = ranked_results.iloc[0]['Model']

print("Best-performing model:", best_model)
```
